```python
import pandas as pd
import numpy as np

sales = pd.read_csv("walmart_master.csv")
```

```python
markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]
sales[markdown_cols] = sales[markdown_cols].fillna(0)
```

```python
sales.isnull().sum()
```

```python
sales.duplicated().sum()
```

```python
sales["Date"] = pd.to_datetime(sales["Date"])
```

```python
sales.dtypes
```

```python
sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month
sales["Month_Name"] = sales["Date"].dt.month_name()
sales["Quarter"] = sales["Date"].dt.quarter
sales["Week"] = sales["Date"].dt.isocalendar().week
```

```python
sales.info()
```

```python
sales.to_csv("walmart_clean.csv", index=False)
```

```python
sales.to_excel("walmart_clean.xlsx", index=False)
```
