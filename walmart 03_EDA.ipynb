```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sales = pd.read_csv("walmart_clean.csv")
sales["Date"] = pd.to_datetime(sales["Date"])
```

```python
plt.figure(figsize=(15,5))
sales.groupby("Date")["Weekly_Sales"].sum().plot()
plt.title("Weekly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

```python
top_store = sales.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
top_store.plot(kind="bar")
plt.title("Top 10 Stores by Sales")
plt.ylabel("Sales")
plt.show()
```

```python
top_dept = sales.groupby("Dept")["Weekly_Sales"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
top_dept.plot(kind="bar")
plt.title("Top Departments")
plt.show()
```

```python
holiday = sales.groupby("IsHoliday")["Weekly_Sales"].sum()
plt.figure(figsize=(6,4))
holiday.plot(kind="bar")
plt.title("Holiday vs Non-Holiday Sales")
plt.show()
```

```python
monthly = sales.groupby("Month_Name")["Weekly_Sales"].sum()

monthly = monthly.reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
])
plt.figure(figsize=(12,5))
monthly.plot()
plt.title("Monthly Sales")
plt.show()
```

```python
sales.groupby("Type")["Weekly_Sales"].sum().plot(kind="bar")
plt.title("Sales by Store Type")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(sales["Weekly_Sales"], bins=50)
plt.title("Distribution of Weekly Sales")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    sales.select_dtypes(include="number").corr(),
    annot=False,
    cmap="coolwarm"
)

plt.show()
```

```python
sales.groupby("Quarter")["Weekly_Sales"].sum().plot(kind="bar")
plt.title("Quarterly Sales")
plt.show()
```

```python
sales.groupby("Year")["Weekly_Sales"].sum().plot(marker="o")
plt.title("Yearly Sales")
plt.show()
```
